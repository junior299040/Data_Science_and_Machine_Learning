**Ex 9 : Créez une fonction qui compte le nombre de fois où le mot "chien" apparaît dans une chaîne de caractères. Encore une fois, ignorez les caractères sur les bords.**

```python
def count_dog(text):
    occur = 0
    for word in range(len(text.lower().split())):
        if text.lower().split()[word] == 'chien':
            occur += 1
    return occur
```

```python
count_dog("Ce chien court plus vite que le chien de l'autre personne !")
```

**Ex 10 : Utilisez des expressions lambda et la fonction filter() pour filtrer les mots d'une liste qui ne commencent pas par la lettre "s". 
Par exemple :**

    seq = ['soupe','chien','salade','chat','super']

**devrait être filtrée ainsi :**

    ['soupe','salade','super']

```python
seq = ['soupe','chien','salade','chat','super']
```

Filter  
Cette fonction prend en entrée une fonction dans laquelle on décrit une condition et une séquence.
Il retient uniquement les éléments de la liste qui respectent la condition décrite

Lambda  
Il s'agit d'un moyen de définir une fonction, elle ne s'exécute qu'une seule fois. Elle évite de définir une fonction avec def pour une simple implémentation.

```python
new_list = list(filter(lambda text : text[0]=='s',seq))
print(new_list)
```

### Problème Final
**Vous conduisez un peu trop vite et un agent de police vous arrête. Ecrivez une fonction qui renvoie un des trois résultats possibles : "Pas de contravention", "Petite contravention", ou "Grosse contravention". 
Si votre vitesse est inférieure ou égale à 60, le résultat est "Pas de contravention". Si la vitesse est comprise entre 61 et 80 inclus, le résultat est "Petite contravention". Si votre vitesse est égale ou supérieure à 81, le résultat est "Grosse contravention". Sauf si c'est votre anniversaire (encodé comme une valeur booléenne dans les paramètres de la fonction) - le jour de votre anniversaire, votre vitesse peut être supérieure de 5 dans tous les cas.**

```python
def caught_speeding(speed, is_birthday):
    uv1,lv2,uv2,lv3,add = 60,61,80,81,5
    def sub_caught_speeding (speed,uv1,lv2,uv2,lv3):
        if all(sub_speed>=0 for sub_speed in [speed, uv1-speed]):
            print("Pas de contravention")
        elif all(sub_speed>=0 for sub_speed in [speed-lv2, uv2-speed]):
            print("Petite contravention")
        elif speed>=lv3:
            print ("Grosse contravention")

    if is_birthday == True:
        sub_caught_speeding(speed,uv1+add,lv2+add,uv2+add,lv3+add)
    else:
        sub_caught_speeding(speed,uv1,lv2,uv2,lv3)
```

```python
caught_speeding(81,True)
```

```python
caught_speeding(81,False)
```
